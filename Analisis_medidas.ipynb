Medidas de Tendencia Central:
1. Media Aritmetica:
Media(μ)= La media es el promedio de un conjunto de datos. Se calcula sumando todos los valores y dividiendo entre el número total de valores.
2. Mediana:
La mediana es el valor central de un conjunto de datos ordenado. Si el número de datos es impar, la mediana es el valor del medio; si es par, es el promedio de los dos valores centrales.
3. Media ponderada:
La media ponderada se utiliza cuando algunos valores tienen más importancia o "peso" que otros. Se calcula multiplicando cada valor por su peso y sumando estos productos, luego se divide entre la suma de los pesos.
4. Moda:
La moda es el valor o los valores que aparecen con mayor frecuencia en un conjunto de datos
